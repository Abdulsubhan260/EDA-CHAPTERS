## Scatter Plot is Seaborn:
this scatter plot shows the relationship between math score and reading score on bases of gender, with each dot representing one item.

The color indicates the item type, so we can compare math score and reading score

It helps us see and compare math score and reading score on the bases of

gender(male/female)


```python
plot=sns.scatterplot(
    data=df,
    x='math score',
    y='reading score',
    hue='gender',
    palette='pastel'
)
plot.set_title('Math Score vs Reading Score',fontweight='bold')
plt.xlabel('math score')
plt.ylabel('reading score')
plt.show()
```

## Cross Tab:

this code shows which students get which marks(like one female ans zero male got zero marks.)

```python
pd.crosstab(index=df['math score'], columns=df['gender'])
```

## pivot_table:
this tabel shows average marks of males/females in math score,reading score and writing scores.

```python
pd.pivot_table(df,index='gender',values=['reading score','math score','writing score'],aggfunc='mean')
```

## Pair Plot:
this code makes a pairplot that compares math score,readin scores and writing scores with the help of diffrent graphs.




```python
sns.pairplot(df,hue='gender')
plt.show()
```

## Bar Chart:
this bar chart shows and compares which gender got higher marks in maths in an easy way

```python
plt.figure(figsize=(8,6))
sns.barplot(data=df,x='gender',y='math score',palette='pastel')
plt.title('Math Score by Gender',fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Math Score')
plt.tight_layout()

plt.show()
```

## BOX plot :
this box plot shows


math scores vary depending on whether students completed a test preparation course or not.


basically this plot shows students who prepare well got arround 65-75 marks


while students who not preapred well lies between 50 and 75 marks.

```python
sns.boxplot(data=df,x='test preparation course',y='math score')
plt.title('Math Score by Test Preparation Course',fontweight='bold')
plt.xlabel('Test Preparation Course')
plt.ylabel('Math Score')
plt.tight_layout()
plt.show()
```

## HISTOGRAM:
in this code i make a histogram to compare maths score on basis of gender.



```python
male = df.loc[df['gender'] == 'male']
female = df.loc[df['gender'] == 'female']

ax = sns.histplot(data=male, x='math score', color='blue', alpha=0.5, label='Male')
ax = sns.histplot(data=female, x='math score', color='red', alpha=0.5, label='Female')

ax.set_xlabel('Math Score', fontsize=10)
ax.set_ylabel('Count', fontsize=10)
ax.set_title('Comparison of Math Score Distribution by Gender', fontsize=12, fontweight='bold')

plt.legend()
plt.show()
```

## Heat map:
in this code i make a heatmap by using seaborn module that shows all correlations between math score,reading score and writing score.


After analysing data i came to know that they all have positive corelation with each other as if one increse all others also increase.

```python

plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(['number']).corr(), annot=True, cmap="icefire")

plt.title("Correlation Heatmap of Student Performance Scores", fontsize=14, fontweight='bold')
plt.show()
```
